## Business-Rule Validation of Missing Values

Missing values were evaluated against workflow status to distinguish expected nulls from potential data-quality issues.

### Support Tickets

`resolution_time_hours` is missing only for tickets with `Open` or `In Progress` status. All `Resolved` and `Closed` tickets have a resolution time.

**Result: Pass.** Missing resolution times are consistent with unresolved ticket states.

### Learning Management

`completed_date` and `completion_time_hours` are missing for `Assigned`, `In Progress`, and `Overdue` records, while all `Completed` records contain both fields.

**Result: Pass.** Missing completion information is consistent with course status.

`assessment_score` contains some null values for records that have not completed or may not yet require assessment.

**Result: Review during processing.** Assessment scores should remain null when an assessment is not applicable or has not yet been completed.

### Onboarding Progress

`completed_date` is missing for `Blocked`, `In Progress`, `Not Started`, and `Overdue` tasks. All `Complete` tasks have a completion date.

**Result: Pass.** Missing completion dates are consistent with incomplete task statuses.

Overall, the identified missing values are primarily workflow-dependent nulls rather than invalid records. They should not be blindly imputed or removed during data cleaning.

```python
print("=" * 60)
print("DATE ORDER VALIDATION")
print("=" * 60)

# Employee: termination should not occur before hire
employee = datasets["Employee"]

hire = pd.to_datetime(employee["DateofHire"], errors="coerce")
termination = pd.to_datetime(employee["DateofTermination"], errors="coerce")

invalid_termination = (
    termination.notna()
    & hire.notna()
    & (termination < hire)
)

print(
    "Employee - termination before hire:",
    invalid_termination.sum()
)


# Learning: completion should not occur before assignment
learning = datasets["Learning Management"]

assigned = pd.to_datetime(
    learning["assigned_date"],
    errors="coerce"
)

completed = pd.to_datetime(
    learning["completed_date"],
    errors="coerce"
)

invalid_learning = (
    completed.notna()
    & assigned.notna()
    & (completed < assigned)
)

print(
    "Learning - completion before assignment:",
    invalid_learning.sum()
)


# Onboarding: completion before due date
onboarding = datasets["Onboarding Progress"]

due = pd.to_datetime(
    onboarding["due_date"],
    errors="coerce"
)

completed = pd.to_datetime(
    onboarding["completed_date"],
    errors="coerce"
)

early_completion = (
    completed.notna()
    & due.notna()
    & (completed < due)
)

print(
    "Onboarding - completed before due date:",
    early_completion.sum()
)
```
